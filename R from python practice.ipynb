```python
import rpy2
import rpy2.rinterface as ri
import rpy2.robjects as ro
from rpy2.robjects.packages import importr
z=ro.r('''
        x=matrix(1:10, 5, 2)
        y=matrix(11:20, 5,2)
        x+y;
        ''')
print(z)
```

```python
base=ro.packages.importr('base')
base.rowSums(z)
# operate the matrix in python 
```
